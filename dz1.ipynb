Написать класс идеального газа для расчета давления, удельного обьема, энтальпии, температуры произвольно заданного газа (например через задание

). Класс помимо расчета состояния газа в точке должен уметь расчитывать изопроцессы.

Примеры использования:

```python
from idealgas import IdealGas

#1. Базовые расчеты параметров газа

#воздух при нормальных условиях
air = IdealGas(
    r=287,      # Газовая постоянная воздуха [Дж/(кг·К)]
    p=101325,   # Давление 1 атм [Па]
    t=293,      # Температура 20°C [К]
    c_p=1005    # Удельная теплоемкость [Дж/(кг·К)]
)

# Расчет производных параметров
print("Базовые параметры воздуха:")
print(f"Удельный объем: {air.find_v():.4f} м³/кг") 
print(f"Энтальпия: {air.find_h()/1000:.2f} кДж/кг") 
print(f"Cv: {air.find_c_v():.2f} Дж/(кг·К)")        
print(f"γ: {air.find_gamma():.4f}")                
```

```python
#2. Изотермический процесс (T=const)
gas = IdealGas(r=287, t=300)  # Задаем только R и T

# Исходное состояние
v1 = 0.8
p1 = gas.r * gas.t / v1

# Процесс сжатия до v2 = 0.5 м³/кг
p2, v2, t = gas.calculation_isothermal_process(v_1=v1, p_1=p1, v_2=0.5)
print("\nИзотермическое сжатие:")
print(f"Начальное давление: {p1:.2f} Па")
print(f"Конечное давление: {p2:.2f} Па")
print(f"Объем изменился с {v1} до {v2:.2f} м³/кг")
print(f"Температура осталась {t:.2f} К")
```

```python
#3. Изоэнтропный (адиабатный) процесс

# Инициализация (азот)
nitrogen = IdealGas(
    r=297,
    c_p=1040,
    p=200000,
    t=300
)

# Исходное состояние
v1 = nitrogen.find_v()

# Адиабатное расширение до p2 = 120 кПа
p2, v2, gamma = nitrogen.calculation_isentropic_process_pv(
    p_1=nitrogen.p,
    v_1=v1,
    p_2=120000
)

print("\nАдиабатное расширение азота:")
print(f"Начальное давление: {nitrogen.p/1000:.1f} кПа")
print(f"Конечное давление: {p2/1000:.1f} кПа")
print(f"Удельный объем изменился с {v1:.4f} до {v2:.4f} м³/кг")
print(f"Показатель адиабаты γ: {gamma:.4f}")
```
